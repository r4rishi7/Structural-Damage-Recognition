```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

PROJECT_ROOT = Path.cwd().parent

TRAIN_X_PATH = (
    PROJECT_ROOT
    / "data"
    / "task2_damage_state_2"
    / "task2_X_train.npy"
)

TRAIN_Y_PATH = (
    PROJECT_ROOT
    / "data"
    / "task2_damage_state_1"
    / "task2_y_train.npy"
)

TEST_X_PATH = (
    PROJECT_ROOT
    / "data"
    / "task2_damage_state_1"
    / "task2_X_test.npy"
)

TEST_Y_PATH = (
    PROJECT_ROOT
    / "data"
    / "task2_damage_state_1"
    / "task2_y_test.npy"
)

X_train = np.load(TRAIN_X_PATH, mmap_mode="r")
y_train = np.load(TRAIN_Y_PATH)

X_test = np.load(TEST_X_PATH, mmap_mode="r")
y_test = np.load(TEST_Y_PATH)

print("X_train:", X_train.shape, X_train.dtype)
print("y_train:", y_train.shape, y_train.dtype)
print("X_test:", X_test.shape, X_test.dtype)
print("y_test:", y_test.shape, y_test.dtype)
```

```python
train_labels = np.argmax(y_train, axis=1)
test_labels = np.argmax(y_test, axis=1)

train_classes, train_counts = np.unique(
    train_labels,
    return_counts=True
)

test_classes, test_counts = np.unique(
    test_labels,
    return_counts=True
)

print("Training class distribution:")
for cls, count in zip(train_classes, train_counts):
    print(f"Class {cls}: {count}")

print("\nTest class distribution:")
for cls, count in zip(test_classes, test_counts):
    print(f"Class {cls}: {count}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 5))

plt.bar(
    ["Class 0", "Class 1"],
    train_counts
)

plt.title("Training Dataset Class Distribution")
plt.xlabel("Class")
plt.ylabel("Number of Images")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))

plt.bar(
    ["Class 0", "Class 1"],
    test_counts
)

plt.title("Test Dataset Class Distribution")
plt.xlabel("Class")
plt.ylabel("Number of Images")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

sample_indices = [0, 100, 1000, 5000]

fig, axes = plt.subplots(2, 2, figsize=(10, 10))

for ax, idx in zip(axes.ravel(), sample_indices):
    ax.imshow(X_train[idx])
    ax.set_title(
        f"Sample {idx} | Class {train_labels[idx]}"
    )
    ax.axis("off")

plt.tight_layout()
plt.show()
```
